# Loan Approval Prediction - Exploratory Data Analysis
This notebook explores the dataset to understand the key factors influencing loan approval.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style
plt.style.use('fivethirtyeight')
sns.set_palette("viridis")
```

```python
# Load data
df = pd.read_csv('../data/train.csv')
df.head()
```

### 1. Basic Info & Missing Values

```python
df.info()
print("\nMissing Values:\n", df.isnull().sum())
```

### 2. Target Variable Distribution

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Loan_Status', data=df)
plt.title('Loan Approval Status Distribution')
plt.show()
```

### 3. Credit History vs Loan Status
Credit History is usually the most important feature.

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Credit_History', hue='Loan_Status', data=df)
plt.title('Impact of Credit History on Loan Approval')
plt.show()
```

### 4. Income Analysis

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Loan_Status', y='ApplicantIncome', data=df)
plt.title('Applicant Income vs Loan Status')
plt.yscale('log') # Log scale due to outliers
plt.show()
```

### 5. Categorical Features Analysis

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

sns.countplot(x='Gender', hue='Loan_Status', data=df, ax=axes[0, 0])
sns.countplot(x='Married', hue='Loan_Status', data=df, ax=axes[0, 1])
sns.countplot(x='Education', hue='Loan_Status', data=df, ax=axes[1, 0])
sns.countplot(x='Property_Area', hue='Loan_Status', data=df, ax=axes[1, 1])

plt.tight_layout()
plt.show()
```
